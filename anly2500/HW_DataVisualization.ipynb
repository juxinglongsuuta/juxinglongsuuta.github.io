## Context
A tip is a monetary incentive given by customers or guests for polite, prompt, and efficient service provided by the staff. The practice of tipping has existed for a long time and has become particularly common in service industries such as hotels and restaurants.

## Objective

Chef's Kitchen is one of the most popular restaurants in San Diego, serving as a one-stop destination for food lovers. The polite and efficient service provided by the restaurant staff often earns them tips from customers. As a Data Analyst for the restaurant, you have been tasked with analyzing the provided data to identify patterns and trends in revenue and tips received from customers across various demographics. Additionally, you are expected to create informative visualizations to convey the insights obtained from the analysis. <br>
<br>
The dataset contains the following features:

- order_id: A unique identifier for each order.
- day: The day of the week when the customer visited the restaurant (Thu, Fri, Sat, or Sun).
- time: The time of day when the customer visited the restaurant (Lunch or Dinner).
- size: The number of people present at the table.
- smoker: Whether the table included smokers (No: The table did not include smokers, Yes: The table included smokers).
- sex: The gender of the person who paid the bill (Male or Female).
- total_bill: The total bill amount in dollars.
- tip: The tip amount in dollars.
