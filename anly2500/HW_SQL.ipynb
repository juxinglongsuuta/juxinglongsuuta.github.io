2. **City** : The name of the city (e.g., Marysville, Perris, Cleveland).

3. **State** : The state where each city is located, represented by the full state name (e.g., Washington, California, Ohio).

4. **Population** : The population of the city, which provides insights into the city’s size.

5. **Latitude (lat)** : The city’s latitude coordinate, specifying its north-south position.

6. **Longitude (lon)** : The city’s longitude coordinate, specifying its east-west position.
